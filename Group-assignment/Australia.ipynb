```python
#G)
gender_counts = df["Sex"].value_counts()
sns.barplot(x=gender_counts.index, y=gender_counts.values, palette="magma")
plt.title("Könsfördelning bland OS-deltagare")
plt.xlabel("Kön")
plt.ylabel("Antalet deltagare")
plt.show
```

```python
#H)
medal_df = df[df["Medal"].notna()]
top_10 = medal_df["Team"].value_counts().head(10)
print(top_10)
sns.barplot(top_10, palette="viridis")
plt.xticks(rotation=45)
plt.tight_layout()
```
