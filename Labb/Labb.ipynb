```python
import matplotlib.pyplot as plt
import pandas as pd
import seaborn as sns

#Uppgift A)

college = pd.read_csv("../Data/College.csv")
```

```python
# Uppgift B)

college2 = pd.read_csv("../Data/College.csv", index_col=0)
college2.head()
```

```python
# Uppgift B)


college3 = college.rename({"Unnamed: 0" : "College"}, axis=1)
college3 = college3.set_index("College")
college = college3
college.head()
```

```python
#Uppgift C)

college.describe()
```

```python
#Uppgift D)

pd.plotting.scatter_matrix(college[["Top10perc", "Apps", "Enroll"]])
```

```python
#Uppgift E)

college.boxplot("Outstate", "Private")
```

```python
# Uppgift F)

college["Elite"]= pd.cut(college["Top10perc"], [0,50,100], labels=["No", "Yes"])
college.boxplot("Outstate","Elite")
```

```python
#Uppgift G)

fig, axes = plt.subplots(2,2, figsize=(16,5))
college["Apps"].plot.hist(bins=30, ax=axes[0,0],color="blue",title="Application")
college["Accept"].plot.hist(bins=50, ax=axes[0,1],color="yellow",title="Accepted")
college["Books"].plot.hist(bins=60, ax=axes[1,0], color="Green", title="Cost of books")
college["Grad.Rate"].plot.hist(bins=70, ax=axes[1,1],color="orange",title="Graduation rate")
plt.tight_layout()
plt.show()
```

```python
# Uppgift H)

college_numeric = college.select_dtypes(include="number")
corr_matrix = college_numeric.corr()
#corr_matrix= corr_matrix[corr_matrix > 0.5]
sns.heatmap(corr_matrix, linewidths=0.07, linecolor="black", cmap="coolwarm")
plt.title("Heatmap over correlation")
plt.show()
```

#### Korrelation mellan olika kolumner
I "heatmapen" ovan så ser vi olika korrelationer i vår data som jag undersöker och förklarar mer här nedan.

```python
fig, axes = plt.subplots(1,3 ,figsize=(15,5))
fig.suptitle("Acceptance into school")
sns.regplot(data=college, x="Apps", y="Accept", ax=axes[0])
sns.regplot(data=college, x="Accept", y="Enroll", ax=axes[1])
sns.regplot(data=college, x="Enroll", y="F.Undergrad", ax=axes[2])
axes[0].set_title("Application vs Accept")
axes[1].set_title("Accept vs Enroll")
axes[2].set_title("Enroll vs Fulltime Undergrad")
plt.tight_layout()
```

#### Antagning
Ju fler studenter som söker en utbildning desto fler studenter blir accepterade och påbörjar en utbildning. En anledning till detta kan vara att skolorna ej har nått sitt maxtak på antalet studenter man kan ta emot. Min teori är att det handlar snarare om att man har möjligheten att ta in studenter än att skolorna är selektiva, då korrelationen mellan dessa punkter förmodligen varit mindre än vad den är.

```python
fig, axes = plt.subplots(1,3 ,figsize=(15,5))
plt.suptitle("Top students choice of school")
sns.regplot(data=college, x="Top10perc", y="Top25perc", ax=axes[0])
sns.regplot(data=college, x="Top10perc", y="Expend", ax=axes[1])
sns.regplot(data=college, x="Top10perc", y="PhD", ax=axes[2])
axes[0].set_title("Top 10% and Top 25% ")
axes[1].set_title("Top 10% and Expenditure")
axes[2].set_title("Top 10% and PhD faculty")
plt.tight_layout()
```

#### Starka akademiska studenter
För studenter som varit bland de 10% och 25% bäst presterande i sin gymnasieutbildning (highschool klass) så ser vi en korrelation med att dem väljer skolor som både har fler fakulteter med en doktorand samt skolor som spenderar mer pengar på att sina elevers lärande. Man kan anta att av denna data att skolan som har flera studenter från både top10/top25 procent är mer selektiv med val av studenter och att det är en prestigefylld akademiskt stark skola. Detta lockar i sin tur fler top-presterande studenter till skolan och det blir en positiv spiral.
